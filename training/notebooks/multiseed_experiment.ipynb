# Preprocessing dan evaluasi light curve ALeRCE ZTF

Notebook ini menjalankan preprocessing secara bertahap dan dapat direview sel demi sel. Unit sampel adalah **OID**, bukan baris detection.

Urutan metodologis:
1. Validasi artefak akuisisi dan audit kelas.
2. Filter `fid=1/2` (g/r), pilih magnitude terkoreksi dengan fallback yang eksplisit.
3. Split stratified pada tingkat OID sebelum statistik preprocessing dihitung.
4. Urutkan setiap sequence berdasarkan `mjd`, lalu `candid` sebagai tie-breaker.
5. Fit normalisasi hanya dari train.
6. Bentuk fitur waktu relatif, delta waktu, magnitude, error, dan one-hot filter.
7. Zero-padding dan binary masking dengan panjang maksimum yang ditentukan dari train.
8. Evaluasi baseline tanpa menggunakan label atau detection dari split lain.

Tidak ada prediksi/probabilitas classifier ALeRCE yang digunakan sebagai fitur.

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import pyarrow.dataset as ds
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix
from alerce_classifier.preprocessing.pipeline import (
    load_labels, split_oids, iter_detection_frames,
    prepare_detection_frame, make_padded_sequences,
)
from alerce_classifier.training.loop import HybridCNNGRU, load_split, make_loader, train_one_epoch, evaluate

DATA_DIR = Path('data/alerce_ztf_lc_sanchezsaez_full/data/full')
OUTPUT_DIR = DATA_DIR / 'preprocessing' / 'notebook_output'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
RANDOM_STATE = 42
TRAIN_FRACTION = 0.70
VALIDATION_FRACTION = 0.15
TEST_FRACTION = 0.15
MAX_LENGTH_QUANTILE = 0.95
# Batas validitas konservatif untuk mencegah nilai sentinel/ekstrem masuk ke model.
MAG_MIN, MAG_MAX = 0.0, 40.0
ERROR_MIN, ERROR_MAX = 0.0, 10.0
FEATURE_COLUMNS = ['time_rel', 'delta_time', 'magnitude', 'magnitude_error', 'filter_g', 'filter_r']
assert abs(TRAIN_FRACTION + VALIDATION_FRACTION + TEST_FRACTION - 1.0) < 1e-9
```

## 1. Validasi akuisisi dan label

`_SUCCESS` wajib ada. Mismatch positif pada `count_mismatches.csv` tidak dihapus karena merupakan audit perubahan database setelah snapshot.

```python
required = [DATA_DIR / '_SUCCESS', DATA_DIR / 'labels_selected.csv', DATA_DIR / 'detections']
missing = [str(p) for p in required if not p.exists()]
if missing:
    raise FileNotFoundError(f'Artefak akuisisi tidak lengkap: {missing}')

labels = load_labels(DATA_DIR / 'labels_selected.csv')
assert labels['oid'].is_unique
print(labels['classMacro'].value_counts())
print(json.loads((DATA_DIR / 'summary.json').read_text()))
```

## 2. Split OID dan pembacaan detection

Split dilakukan sebelum normalisasi. Semua detection dari satu OID selalu berada pada split yang sama. Pembacaan menggunakan PyArrow scanner agar tidak seluruh dataset masuk RAM.

```python
labels = split_oids(labels, RANDOM_STATE, VALIDATION_FRACTION, TEST_FRACTION)
train_ids = labels.loc[labels['split'] == 'train', 'oid'].tolist()
validation_ids = labels.loc[labels['split'] == 'validation', 'oid'].tolist()
test_ids = labels.loc[labels['split'] == 'test', 'oid'].tolist()
print(pd.crosstab(labels['split'], labels['classMacro'], normalize='index'))

DETECTION_COLUMNS = ['oid', 'candid', 'mjd', 'fid', 'magpsf', 'sigmapsf', 'magpsf_corr', 'sigmapsf_corr', 'corrected']
parts = list(iter_detection_frames(DATA_DIR / 'detections', labels, batch_size=200_000))
detections = pd.concat(parts, ignore_index=True)
print(f'detections g/r: {len(detections):,}; OID: {detections.oid.nunique():,}')
```

## 3. Pembersihan fotometri dan pengurutan kronologis

Pengurutan dilakukan **sebelum** fitur sequence dibuat: `mjd` menaik, kemudian `candid` menaik jika dua detection memiliki MJD sama. `time_rel` dihitung dari MJD pertama pada OID tersebut, sedangkan `delta_time` adalah selisih terhadap detection sebelumnya. Detection dengan MJD, magnitude, atau error yang tidak valid dibuang secara eksplisit dan jumlahnya dicatat.

Kebijakan magnitude: gunakan `magpsf_corr` hanya ketika `corrected == True`, finite, dan berada pada rentang fisik; jika tidak, gunakan `magpsf` yang juga harus valid.

```python
detections = detections.sort_values(['oid', 'mjd', 'candid'], kind='mergesort').reset_index(drop=True)
corrected_ok = (
    detections['corrected'].eq(True)
    & detections['magpsf_corr'].between(MAG_MIN, MAG_MAX, inclusive='both')
    & detections['sigmapsf_corr'].between(ERROR_MIN, ERROR_MAX, inclusive='both')
)
detections['magnitude'] = detections['magpsf'].where(~corrected_ok, detections['magpsf_corr'])
detections['magnitude_error'] = detections['sigmapsf'].where(~corrected_ok, detections['sigmapsf_corr'])
valid = (
    detections['mjd'].between(0, np.inf)
    & detections['magnitude'].between(MAG_MIN, MAG_MAX)
    & detections['magnitude_error'].between(ERROR_MIN, ERROR_MAX)
)
quality_report = {
    'rows_before_quality_filter': int(len(detections)),
    'rows_after_quality_filter': int(valid.sum()),
    'rows_removed': int((~valid).sum()),
    'fid3_removed_before_processing': 'yes',
}
detections = detections.loc[valid].copy()
detections['time_rel'] = detections['mjd'] - detections.groupby('oid')['mjd'].transform('first')
detections['delta_time'] = detections.groupby('oid')['mjd'].diff().fillna(0.0)
detections['filter_g'] = detections['fid'].eq(1).astype('float32')
detections['filter_r'] = detections['fid'].eq(2).astype('float32')
assert (detections.groupby('oid')['mjd'].diff().dropna() >= 0).all()
print(quality_report)
```

## 4.1 Audit duplikasi, missing value, dan time window

Validasi akuisisi sebelumnya sudah memeriksa keunikan `(oid, candid)` per batch. Sel ini mengulang audit pada data g/r yang akan masuk model. Time window di sini adalah durasi kalender observasi per OID, `max(mjd)-min(mjd)`, bukan panjang sequence atau `MAX_LENGTH`.

```python
null_report = detections.isna().sum()
duplicate_count = int(detections.duplicated(['oid', 'candid']).sum())
time_windows = detections.groupby('oid')['mjd'].agg(['min', 'max', 'count'])
time_windows['duration_days'] = time_windows['max'] - time_windows['min']
time_window_summary = time_windows['duration_days'].describe(percentiles=[0.05, 0.25, 0.5, 0.75, 0.95, 0.99]).to_dict()
print('Missing values setelah filter g/r:')
print(null_report[null_report > 0])
print('Duplicate (oid, candid):', duplicate_count)
print('Time window summary (days):', time_window_summary)
time_windows.to_csv(OUTPUT_DIR / 'time_windows_by_oid.csv')
json_report = {'null_values': {str(k): int(v) for k, v in null_report.items()}, 'duplicate_oid_candid': duplicate_count, 'time_window_days': time_window_summary}
(OUTPUT_DIR / 'data_quality_and_time_window.json').write_text(json.dumps(json_report, indent=2), encoding='utf-8')
```

## 4. Normalisasi train-only dan sequence variable-length

Mean/std dihitung hanya dari baris detection milik OID train. Validation dan test ditransformasi menggunakan statistik tersebut. Sequence tetap variable-length pada tahap ini, sehingga urutan kronologis dapat diperiksa sebelum padding.

```python
train_oid_set = set(train_ids)
train_rows = detections[detections['oid'].isin(train_oid_set)]
normalization_columns = ['time_rel', 'delta_time', 'magnitude', 'magnitude_error']
normalization = {}
for column in normalization_columns:
    mean = float(train_rows[column].mean())
    std = float(train_rows[column].std(ddof=0))
    normalization[column] = {'mean': mean, 'std': std if std > 0 else 1.0}
    detections[column] = (detections[column] - mean) / normalization[column]['std']
(OUTPUT_DIR / 'normalization.json').write_text(json.dumps(normalization, indent=2), encoding='utf-8')

sequences = {}
for oid, group in detections.groupby('oid', sort=False):
    sequences[oid] = group[FEATURE_COLUMNS].to_numpy(dtype='float32')
train_lengths = [len(sequences[oid]) for oid in train_ids if oid in sequences]
MAX_LENGTH = max(1, int(np.ceil(np.quantile(train_lengths, MAX_LENGTH_QUANTILE))))
print(f'MAX_LENGTH dari quantile train {MAX_LENGTH_QUANTILE}: {MAX_LENGTH}')
```

## 5. Zero-padding dan binary masking

Model menerima tensor `(jumlah_oid, MAX_LENGTH, jumlah_fitur)`. Nilai setelah panjang sequence asli diisi nol. `mask=1` hanya untuk observasi asli dan `mask=0` untuk padding. Mask bukan fitur yang dihitung dari masa depan; ia hanya menandai keberadaan observasi. `MAX_LENGTH` ditetapkan pada P95 sequence train (1.088 timestep) agar penggunaan memori tetap terkendali. Sequence yang lebih panjang dipotong setelah timestep ke-1.088, sehingga prefix kronologis tetap dipertahankan. Jumlah truncation dilaporkan per split dan per kelas.

```python
def pad_sequences(oids):
    selected = [oid for oid in oids if oid in sequences]
    x = np.zeros((len(selected), MAX_LENGTH, len(FEATURE_COLUMNS)), dtype='float32')
    mask = np.zeros((len(selected), MAX_LENGTH), dtype='float32')
    truncated = 0
    for index, oid in enumerate(selected):
        sequence = sequences[oid]
        if len(sequence) > MAX_LENGTH:
            sequence = sequence[:MAX_LENGTH]
            truncated += 1
        length = len(sequence)
        x[index, :length] = sequence
        mask[index, :length] = 1.0
    return selected, x, mask, truncated

class_to_index = {'Periodic': 0, 'Stochastic': 1, 'Transient': 2}
index_to_class = {value: key for key, value in class_to_index.items()}
padded = {}
truncation_report = []
for split_name, split_ids in [('train', train_ids), ('validation', validation_ids), ('test', test_ids)]:
    selected, x, mask, truncated = pad_sequences(split_ids)
    y = labels.set_index('oid').loc[selected, 'classMacro'].map(class_to_index).to_numpy(dtype='int64')
    padded[split_name] = {'oids': selected, 'x': x, 'mask': mask, 'y': y}
    np.savez_compressed(OUTPUT_DIR / f'{split_name}_padded.npz', x=x, mask=mask, y=y)
    split_labels = labels.set_index('oid').loc[selected, 'classMacro']
    truncated_by_class = labels.set_index('oid').loc[
        [oid for oid in selected if len(sequences[oid]) > MAX_LENGTH], 'classMacro'
    ].value_counts().to_dict()
    truncation_report.append({'split': split_name, 'total_truncated': truncated, 'truncated_by_class': truncated_by_class})
    print(split_name, x.shape, mask.sum(axis=1).mean(), 'truncated:', truncated, truncated_by_class)
(OUTPUT_DIR / 'truncation_report.json').write_text(json.dumps(truncation_report, indent=2), encoding='utf-8')
```

## 6. Training weighted cross entropy dan evaluasi

Class weight dihitung dari OID train saja. Training memakai weighted cross entropy; validation dan test tetap memakai distribusi alami tanpa sampler seimbang.

```python
classes = np.array([0, 1, 2])
train_y = padded['train']['y']
weights = compute_class_weight(class_weight='balanced', classes=classes, y=train_y)
class_weights = {index_to_class[int(key)]: float(value) for key, value in zip(classes, weights)}
(OUTPUT_DIR / 'class_weights.json').write_text(json.dumps(class_weights, indent=2), encoding='utf-8')
print(class_weights)
```

## Catatan padding dan kronologi

- Sorting dilakukan per OID dengan `mjd` menaik; `candid` hanya tie-breaker deterministik.
- `time_rel = mjd - mjd_pertama` mempertahankan sampling tidak seragam.
- `delta_time` detection pertama bernilai 0, detection berikutnya adalah selisih MJD aktual.
- Padding dilakukan setelah sorting dan feature engineering, bukan sebelum.
- Mask bernilai 1 untuk data aktual dan 0 untuk padding; saat memakai GRU/CNN, mask harus diteruskan atau digunakan untuk mengabaikan timestep padding.
- Untuk robustness early-detection, potong sequence test secara kronologis sebelum padding dan jangan menghitung ulang statistik dari prefix test.

## 7. Konfigurasi eksperimen final

Jalankan blok-blok di bawah secara berurutan setelah preprocessing selesai. Eksperimen utama menggunakan maksimum 50 epoch dan early stopping berdasarkan validation macro-F1. Multi-seed, baseline, dan prefix evaluation dapat diaktifkan sesuai kebutuhan penelitian.

```python
RUN_FINAL_EXPERIMENTS = True
RUN_BASELINES = True
RUN_MULTI_SEED = True
RUN_PREFIX_EVALUATION = True
EXPERIMENT_SEEDS = [42, 123, 2024]
EPOCHS = 50
PATIENCE = 7
BATCH_SIZE = 64
LEARNING_RATE = 1e-3
print(f'epochs={EPOCHS}, patience={PATIENCE}, batch_size={BATCH_SIZE}')
```

### 7.1 Memuat tensor dan menyiapkan device


```python
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import importlib
import alerce_classifier.training.loop as training_module
training_module = importlib.reload(training_module)
from alerce_classifier.training.loop import (
    CNNClassifier, GRUClassifier, evaluate, evaluate_prefixes,
    set_seed, train_model,
)
train_tensors = load_split(OUTPUT_DIR / 'train_padded.npz')
validation_tensors = load_split(OUTPUT_DIR / 'validation_padded.npz')
test_tensors = load_split(OUTPUT_DIR / 'test_padded.npz')
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
seeds = EXPERIMENT_SEEDS if RUN_MULTI_SEED else [42]
print(f'Device: {device}')
print(f'Seeds: {seeds}')
```

### 7.2 Menentukan model yang dijalankan


```python
model_factories = {'hybrid_cnn_gru': HybridCNNGRU}
if RUN_BASELINES:
    model_factories.update({'cnn': CNNClassifier, 'gru': GRUClassifier})
print('Models:', list(model_factories))
if not RUN_FINAL_EXPERIMENTS:
    print('Training belum dijalankan. Ubah RUN_FINAL_EXPERIMENTS=True untuk memulai.')
```

### 7.3 Training dengan checkpoint terbaik dan early stopping


```python
experiment_runs = {}
if RUN_FINAL_EXPERIMENTS:
    for model_name, factory in model_factories.items():
        for seed in seeds:
            print(f'\nTraining {model_name} (seed={seed})', flush=True)
            set_seed(seed)
            train_loader = make_loader(*train_tensors, batch_size=BATCH_SIZE, shuffle=True)
            validation_loader = make_loader(*validation_tensors, batch_size=BATCH_SIZE)
            model = factory(feature_count=len(FEATURE_COLUMNS)).to(device)
            key = f'{model_name}_seed{seed}'
            checkpoint = OUTPUT_DIR / f'{key}_best.pt'
            history, selection = train_model(
                model, train_loader, validation_loader, weights, device,
                epochs=EPOCHS, patience=PATIENCE, learning_rate=LEARNING_RATE,
                checkpoint_path=checkpoint,
            )
            experiment_runs[key] = {'model': model, 'history': history, 'selection': selection}
            (OUTPUT_DIR / f'{key}_history.json').write_text(json.dumps(history, indent=2), encoding='utf-8')
else:
    experiment_runs = {}
```

### 7.4 Evaluasi test satu kali setelah training


```python
all_results = {}
if RUN_FINAL_EXPERIMENTS:
    test_loader = make_loader(*test_tensors, batch_size=BATCH_SIZE)
    for key, run in experiment_runs.items():
        test_metrics = evaluate(run['model'], test_loader, device)
        all_results[key] = {'selection': run['selection'], 'test': test_metrics}
        print(key, {metric: test_metrics[metric] for metric in ('accuracy', 'balanced_accuracy', 'macro_f1')})
    (OUTPUT_DIR / 'experiment_results.json').write_text(json.dumps(all_results, indent=2), encoding='utf-8')
```

### 7.5 Grafik training dan confusion matrix


```python
if RUN_FINAL_EXPERIMENTS:
    for key, run in experiment_runs.items():
        history = run['history']
        epochs = [item['epoch'] for item in history]
        fig, axes = plt.subplots(1, 2, figsize=(11, 4))
        axes[0].plot(epochs, [item['train_loss'] for item in history], label='train loss')
        axes[0].plot(epochs, [item['loss'] for item in history], label='validation loss')
        axes[0].set(xlabel='Epoch', ylabel='Loss', title=f'{key} loss')
        axes[0].legend(); axes[0].grid(alpha=.3)
        axes[1].plot(epochs, [item['macro_f1'] for item in history], label='validation macro-F1')
        axes[1].plot(epochs, [item['accuracy'] for item in history], label='validation accuracy')
        axes[1].set(xlabel='Epoch', ylabel='Score', title=f'{key} metrics')
        axes[1].legend(); axes[1].grid(alpha=.3)
        fig.tight_layout(); fig.savefig(OUTPUT_DIR / f'{key}_training_curves.png', dpi=150); plt.show()
        plt.figure(figsize=(5, 4))
        sns.heatmap(all_results[key]['test']['confusion_matrix'], annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
        plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title(f'{key} test confusion matrix')
        plt.tight_layout(); plt.savefig(OUTPUT_DIR / f'{key}_confusion_matrix.png', dpi=150); plt.show()
```

### 7.6 Evaluasi prefix early detection (opsional)


```python
if RUN_FINAL_EXPERIMENTS and RUN_PREFIX_EVALUATION:
    key = f'hybrid_cnn_gru_seed{seeds[0]}'
    prefix_metrics = evaluate_prefixes(
        experiment_runs[key]['model'], test_tensors[0], test_tensors[1], test_tensors[2],
        [0.1, 0.25, 0.5, 0.75, 1.0], device,
    )
    (OUTPUT_DIR / f'{key}_prefix_metrics.json').write_text(json.dumps(prefix_metrics, indent=2), encoding='utf-8')
    print(json.dumps(prefix_metrics, indent=2))
```
